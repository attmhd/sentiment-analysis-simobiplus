```python
import pandas as pd
from google_play_scraper import reviews, Sort
```

Mengambil Data Reviewpada Aplikasi [**SimobiPlus**](https://play.google.com/store/apps/details?id=com.simas.mobile.SimobiPlus&hl=id)

```python
APK_ID = 'com.simas.mobile.SimobiPlus'
JUMLAH_DATA = 10000
LANG = 'id'

scrapreview = reviews(
    APK_ID,          
    lang=LANG,             
    country=LANG,     
    sort=Sort.NEWEST, 
    count=JUMLAH_DATA             
)
```

```python
review_df = pd.DataFrame(scrapreview[0])
review_df.head()
```

Ekspor Dataset

```python
OUTPUT_PATH = '/kaggle/working/simobi_reviews.csv'

review_df.to_csv(OUTPUT_PATH, index=None, header=True)
```
